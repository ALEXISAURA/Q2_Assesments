```python
#I'ma let the user choose
print("Choose an operation")
print("1-addition")
print("2-subtraction")
print("3-multiplication")
print("4-division")

choice = input("Enter chosen operation (1/2/3/4): ")

#Input they're wanted number
n1 = int(input("Enter first number"))
n2 = int(input("Enter second number"))

#Time to solve
if choice == '1':
  print("Result:", n1+n2)

elif choice == '2':
  print("Result:", n1-n2)

elif choice == '3':
  print("Result:", n1*n2)

elif choice == '4':

   if n2 != 0:
    print("Result:", n1/n2)

   else:
    print("Not appropriate please use a non zero number")

else:
  print("Try again! PLease choose between the 4 operations given")






```
