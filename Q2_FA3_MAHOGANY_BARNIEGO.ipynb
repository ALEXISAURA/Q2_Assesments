```python
def show_steps(steps_list):
    print("\n--- Daily Steps Log ---")
    if not steps_list:
        print("No step entries recorded.")
        return
    for index, steps in enumerate(steps_list):
        print(f"Day {index + 1} [Index {index}]: {steps:,} steps")


def find_steps(steps_list, target_steps):
    matches = []
    for i in range(len(steps_list)):
        if steps_list[i] == target_steps:
            matches.append(i)

    if matches:
        print(f"Found {target_steps:,} steps at index/indices: {matches}")
    else:
        print(f"No day had an exact count of {target_steps:,} steps.")


def compute_stats(steps_list):
    if not steps_list:
        print("No data available for computation.")
        return
    total = sum(steps_list)
    average = total / len(steps_list)
    highest = max(steps_list)
    lowest = min(steps_list)
    print(f"Total Steps:   {total:,}")
    print(f"Average Steps: {average:,.2f}")
    print(f"Highest Day:   {highest:,}")
    print(f"Lowest Day:    {lowest:,}")


def update_steps(steps_list, index, new_steps):
    if 0 <= index < len(steps_list):
        previous = steps_list[index]
        steps_list[index] = new_steps
        print(f"Updated Index {index} from {previous:,} to {new_steps:,} steps.")
    else:
        print("Out-of-range index! Nothing was updated.")


def run_tracker():
    daily_steps = [6420, 8150, 5300, 9480, 7210]

    while True:
        print("\n=== Daily Step Counter Analyzer ===")
        print("Select the number of the operation you want to perform:")
        print("1. Display Steps")
        print("2. Search Steps")
        print("3. Compute Step Statistics")
        print("4. Update Steps")
        print("5. Display Updated Steps")
        print("6. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            show_steps(daily_steps)
        elif choice == "2":
            target = int(input("Enter exact step count to search: "))
            find_steps(daily_steps, target)
        elif choice == "3":
            compute_stats(daily_steps)
        elif choice == "4":
            show_steps(daily_steps)
            idx = int(input("Enter index to update: "))
            updated_val = int(input("Enter new step count: "))
            update_steps(daily_steps, idx, updated_val)
        elif choice == "5":
            show_steps(daily_steps)
        elif choice == "6":
            print("Closing Step Counter Analyzer. Keep moving!")
            break
        else:
            print("Invalid option. Please choose 1 to 6.")


if __name__ == "__main__":
    run_tracker()
```
